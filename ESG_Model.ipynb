```python
import pandas as pd
import numpy as np

class ESGModel:
    def _init_(self):
        # Load your base ESG dataset (make sure final_dataset.csv is in the same folder)
        self.data = pd.read_csv("final_dataset.csv")
    
    def predict(self, df_new):
        """
        Predict ESG score for new company data.
        For demo, we simulate ESG scores using random values.
        """
        df = df_new.copy()
        np.random.seed(42)
        df['ESG_Score'] = np.round(np.random.uniform(0, 100, size=len(df)), 2)
        return df

    def green_status(self, df_combined, sector_rules):
        """
        Determine Green/Not Green based on ESG score and years.
        """
        df = df_combined.copy()
        df['Years_Green'] = np.random.randint(1, 10, size=len(df))
        green_list = []

        for idx, row in df.iterrows():
            sector = row['Sector']
            required_years = sector_rules.get(sector, 2)
            if row['Years_Green'] >= required_years and row['ESG_Score'] >= 50:
                green_list.append("Green")
            else:
                green_list.append("Not Green")
        
        df['Final_Green_Status'] = green_list
        return df[['Company_Name', 'Sector', 'ESG_Score', 'Years_Green', 'Final_Green_Status']]
```
